```python
import pandas as pd
import re
```

```python
lines = open('../test_outputs/file_changes.txt', 'r')
lines = [line.replace('\n', '') for line in lines]
lines
```

```python
pattern = r"^(?P<artists>[\w\s&,]+?)\s*-\s*(?P<title>.+)$"
matched = []
for line in lines:
    match = re.match(pattern, line)
    if match:
        matched.append((line,match.group('artists'), match.group("title")))
    else:
        matched.append((line, None, None))
matched
```

```python
print(sum(1 for _, match, _ in matched if match), len(matched), sum(1 for _, match, _ in matched if match)/len(matched))
```
